```python
import zipfile
import os
import shutil

# Replace 'your_uploaded_file.zip' with the actual name of your uploaded zip file
zip_file_path = 'python_client.zip'

# Create a directory to extract the files into (optional)
extract_dir = 'extracted_files'
os.makedirs(extract_dir, exist_ok=True)

# Unzip the file
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"File '{zip_file_path}' unzipped successfully to '{extract_dir}'.")
# move out the python_client folder from extracted folders outside
# Define the source and destination paths
source_path = os.path.join(extract_dir, 'python_client')
destination_path = '.' # Move to the current directory (where the notebook is)

# Move the directory
if os.path.exists(source_path):
    shutil.move(source_path, destination_path)
    print(f"Moved '{source_path}' to '{destination_path}'.")
else:
    print(f"Source directory '{source_path}' not found.")

# Now change the directory to python_client
os.chdir('python_client/')
print(f"Changed current directory to: {os.getcwd()}")

# List files in the current directory to verify
print("Files in the current directory:")
print(os.listdir())
print(os.listdir())
#asdas
```
